# Fig. 2: Halo 2's DF under the four particle selections

Reads `Halo_Results/Halo_02/df/raw/`. Writes `fig2_<selection>.dat` to `output/`.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().resolve().parents[1]))  # AurigaDFs/, so the package imports

import numpy as np
import pandas as pd

from Empirical_DF_Construction.io.results_paths import df_raw_dir
from Empirical_DF_Construction.paper_figures.common import EPS_HI, EPS_LO, REF_HALO, out_dir

SELECTIONS = {
    "without_hubble": "df_raw_without_hubble_flow.csv",
    "with_hubble": "df_raw_with_hubble_flow.csv",
    "fixed_point": "df_raw_fixed_point.csv",
    "radial": "df_naive_radial_R200m.csv",
}
```

```python
out = out_dir()
for name, csv in SELECTIONS.items():
    d = pd.read_csv(df_raw_dir(REF_HALO) / csv)
    d = d[np.isfinite(d["log10f"]) & d["eps_hat"].between(EPS_LO, EPS_HI)]
    np.savetxt(out / f"fig2_{name}.dat", d[["eps_hat", "log10f"]].to_numpy(), fmt="%.8g",
               header="eps_hat log10f")
    print(f"{name:15s} {len(d)} points")
```
